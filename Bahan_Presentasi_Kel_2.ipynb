```python

# ============================================================
# 1. IMPORT LIBRARY
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.metrics import davies_bouldin_score
from sklearn.decomposition import PCA

sns.set_style("whitegrid")


# ============================================================
# 2. IMPORT DATASET WINE CLUSTERING
# ============================================================

# Menghubungkan Google Colab dengan file dataset
from google.colab import files

uploaded = files.upload()

# Membaca dataset CSV yang diupload
wine = pd.read_csv("wine-clustering.csv")

print("Ukuran Dataset:", wine.shape)

print("\nInformasi Dataset:")
print(wine.info())

print("\nLima Data Pertama:")
display(wine.head())



# ============================================================
# 3. PENGECEKAN DATA
# ============================================================

print("\nJumlah Missing Values:")
print(wine.isnull().sum())

print("\nJumlah Data Duplikat:")
print(wine.duplicated().sum())

print("\nStatistik Deskriptif:")
display(wine.describe())


# ============================================================
# 4. NORMALISASI DATA MENGGUNAKAN MINMAX SCALER
# ============================================================

# Menggunakan seluruh variabel numerik
features = wine.select_dtypes(include=np.number).columns.tolist()

print("\nVariabel yang Digunakan:")
print(features)

scaler = MinMaxScaler()

wine_scaled = scaler.fit_transform(wine[features])

print("\nData Setelah Normalisasi:")

display(pd.DataFrame(
    wine_scaled,
    columns=features
).head())


# ============================================================
# 5. ELBOW METHOD
# ============================================================

sse = []

for k in range(1, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(wine_scaled)

    sse.append(kmeans.inertia_)

print("\nNilai SSE Setiap Jumlah Cluster:")

for k, nilai in enumerate(sse, start=1):
    print(f"Jumlah Cluster {k}: SSE = {nilai:.4f}")


# ============================================================
# 6. EVALUASI DAVIES-BOULDIN INDEX (DBI)
# ============================================================

dbi_values = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(wine_scaled)

    dbi = davies_bouldin_score(
        wine_scaled,
        labels
    )

    dbi_values.append(dbi)

    print(f"Jumlah Cluster {k}: DBI = {dbi:.4f}")


# Menentukan jumlah cluster dengan DBI terkecil
best_k_dbi = range(2, 11)[np.argmin(dbi_values)]

print("\nJumlah Cluster Terbaik Berdasarkan DBI:", best_k_dbi)

print("Nilai DBI Terkecil:", round(min(dbi_values), 4))


# ============================================================
# 7. MEMBENTUK MODEL K-MEANS
# ============================================================

# Menggunakan 3 cluster sebagai jumlah cluster awal
# Selanjutnya dapat dievaluasi menggunakan Elbow dan DBI.

optimal_k = 3

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

labels = kmeans.fit_predict(wine_scaled)

# Menambahkan hasil cluster ke dataset asli
wine["Cluster"] = labels

# Menghitung DBI untuk model dengan 3 cluster
dbi_final = davies_bouldin_score(
    wine_scaled,
    labels
)

print("\nDBI Model K-Means dengan 3 Cluster:",
      round(dbi_final, 4))


# ============================================================
# 8. VISUALISASI ELBOW METHOD
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    sse,
    marker="o",
    linewidth=2
)

plt.title("Elbow Method - Wine Clustering")
plt.xlabel("Jumlah Cluster (K)")
plt.ylabel("Sum of Squared Error (SSE)")
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()


# ============================================================
# 9. VISUALISASI DAVIES-BOULDIN INDEX
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 11),
    dbi_values,
    marker="o",
    color="red",
    linewidth=2
)

# Menandai DBI terkecil
plt.scatter(
    best_k_dbi,
    min(dbi_values),
    color="blue",
    s=100,
    label=f"DBI Terbaik (K={best_k_dbi})"
)

# Menandai K=3 yang digunakan dalam model
plt.axvline(
    x=3,
    color="green",
    linestyle="--",
    label="K = 3 (Model)"
)

plt.title("Evaluasi K-Means Menggunakan DBI")
plt.xlabel("Jumlah Cluster (K)")
plt.ylabel("Davies-Bouldin Index")
plt.xticks(range(2, 11))
plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# 10. VISUALISASI CLUSTER MENGGUNAKAN PCA
# ============================================================

# Mengurangi dimensi data menjadi 2 komponen
pca = PCA(n_components=2)

wine_pca = pca.fit_transform(wine_scaled)

pca_df = pd.DataFrame({
    "PCA 1": wine_pca[:, 0],
    "PCA 2": wine_pca[:, 1],
    "Cluster": labels
})

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x="PCA 1",
    y="PCA 2",
    hue="Cluster",
    palette="Set1",
    s=70
)

plt.title("Visualisasi Hasil K-Means Clustering (PCA)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Cluster")
plt.grid(True)
plt.show()

print(
    "\nTotal Variasi yang Dijelaskan PCA:",
    round(sum(pca.explained_variance_ratio_) * 100, 2),
    "%"
)


# ============================================================
# 11. VISUALISASI JUMLAH ANGGOTA SETIAP CLUSTER
# ============================================================

cluster_counts = wine["Cluster"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=cluster_counts.index.astype(str),
    y=cluster_counts.values,
    hue=cluster_counts.index.astype(str),
    palette="Set2",
    legend=False
)

plt.title("Jumlah Anggota pada Setiap Cluster")
plt.xlabel("Cluster")
plt.ylabel("Jumlah Data")
plt.show()

print("\nJumlah Anggota Cluster:")
print(cluster_counts)


# ============================================================
# 12. ANALISIS RATA-RATA SETIAP CLUSTER
# ============================================================

cluster_mean = wine.groupby("Cluster")[features].mean()

print("\nRata-rata Variabel pada Setiap Cluster:")
display(cluster_mean)


# ============================================================
# 13. VISUALISASI KARAKTERISTIK CLUSTER
# ============================================================

# Menghitung rata-rata data yang telah dinormalisasi
scaled_df = pd.DataFrame(
    wine_scaled,
    columns=features
)

scaled_df["Cluster"] = labels

cluster_profile = scaled_df.groupby("Cluster")[features].mean()

plt.figure(figsize=(14, 7))

sns.heatmap(
    cluster_profile,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Heatmap Karakteristik Setiap Cluster")
plt.xlabel("Variabel")
plt.ylabel("Cluster")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 14. KESIMPULAN HASIL EVALUASI
# ============================================================

print("\n==========================================")
print("KESIMPULAN HASIL CLUSTERING")
print("==========================================")

print("Jumlah observasi:", len(wine))

print("Jumlah variabel yang digunakan:", len(features))

print("Jumlah cluster yang digunakan:", optimal_k)

print("Nilai DBI model:", round(dbi_final, 4))

print("Jumlah cluster terbaik berdasarkan DBI:", best_k_dbi)

print("Nilai DBI terkecil:", round(min(dbi_values), 4))


print("\nINTERPRETASI DBI")

print("Nilai Davies-Bouldin Index (DBI) digunakan untuk")
print("mengevaluasi kualitas hasil clustering.")

print("Nilai DBI yang lebih kecil menunjukkan hasil clustering")
print("yang lebih baik karena cluster cenderung lebih kompak")
print("dan memiliki pemisahan yang lebih jelas.")

if dbi_final < 1:
    print("\nKualitas clustering dengan 3 cluster relatif baik.")
elif dbi_final < 2:
    print("\nKualitas clustering dengan 3 cluster masih perlu dievaluasi.")
else:
    print("\nKualitas clustering dengan 3 cluster perlu diperbaiki.")


print("\nPERBANDINGAN METODE")

print("Elbow Method digunakan untuk menentukan jumlah cluster")
print("berdasarkan perubahan nilai Sum of Squared Error (SSE).")

print("Sementara itu, DBI digunakan untuk mengevaluasi kualitas")
print("cluster berdasarkan kekompakan dan pemisahan antarcluster.")

print("Kedua metode sebaiknya digunakan secara bersama-sama")
print("agar pemilihan jumlah cluster lebih meyakinkan.")


if best_k_dbi == optimal_k:

    print("\nHasil DBI menunjukkan jumlah cluster terbaik")
    print("yang sama dengan model, yaitu", optimal_k, "cluster.")

else:

    print("\nHasil DBI dan model menunjukkan jumlah cluster berbeda.")

    print("Model K-Means menggunakan", optimal_k, "cluster.")

    print("DBI menunjukkan nilai terbaik pada", best_k_dbi, "cluster.")

    print("Oleh karena itu, perlu dipertimbangkan kembali hasil")
    print("visualisasi Elbow Method dan karakteristik cluster.")


print("\nCATATAN")

print("DBI yang lebih rendah menunjukkan kualitas clustering")
print("yang lebih baik secara relatif.")

print("Penentuan jumlah cluster sebaiknya mempertimbangkan")
print("hasil Elbow Method, DBI, dan interpretasi karakteristik cluster.")
```
